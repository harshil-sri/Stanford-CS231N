## Sequence Modeling Paradigms

Standard feedforward networks map one fixed-size input vector to one fixed-size output vector. Sequence models drop that constraint so you can work with variable-length inputs and outputs.

Common sequence setups:
* **One-to-many:** Single input to a sequence output (e.g., Image Captioning).
* **Many-to-one:** Sequence input to a single output (e.g., Video Action Recognition, Sentiment Analysis).
* **Many-to-many (synced):** Synchronized sequence in to sequence out (e.g., frame-level video classification).
* **Many-to-many (async / Seq2Seq):** Encoder-decoder setup where a many-to-one network encodes the input into a vector, then a one-to-many network decodes that vector into a new sequence (e.g., Machine Translation).

## Training dynamics and gradient flow

RNNs train using Backpropagation Through Time (BPTT). This runs a forward pass through the full sequence to get the total loss, then backpropagates backward through all time steps.

Truncated BPTT splits long sequences into smaller chunks. Hidden states pass forward across chunks indefinitely, but you only compute and backpropagate gradients over a set number of recent steps to save memory and compute.

Multiplying by $W_{hh}^T$ over and over during backpropagation causes numerical stability issues:
* **Exploding gradients**: If the largest singular value of $W_{hh}$ is over 1, gradients grow exponentially on long sequences.
* **Vanishing gradients**: If the largest singular value is under 1, or when gradients repeatedly pass through the derivative of the squashing $\tanh$ function, signals from earlier steps decay to zero.

You can fix exploding gradients with gradient clipping, which rescales the gradient vector whenever its L2 norm exceeds a set threshold:


```python
# Gradient clipping implementation
grad_norm = np.linalg.norm(grad)
if grad_norm > threshold:
    grad *= (threshold / grad_norm)
```

## Long Short-Term Memory (LSTM) Architecture

LSTMs fix vanishing gradients by adding a dedicated cell state $c_t$ alongside the hidden state $h_t$. Four internal gates control what enters and leaves the cell state:

$\begin{pmatrix} i \\ f \\ o \\ g \end{pmatrix} = \begin{pmatrix} \sigma \\ \sigma \\ \sigma \\ \tanh \end{pmatrix} W \begin{pmatrix} h_{t-1} \\ x_t \end{pmatrix}, \quad c_t = f \odot c_{t-1} + i \odot g, \quad h_t = o \odot \tanh(c_t)$

The forget gate $f$ clears old memory from the cell state. The input gate $i$ and candidate state $g$ write new information. The output gate $o$ decides how much of the cell state to pass to the hidden state.

Gradients flow back along the cell state $c_t$ through elementwise multiplication by $f$ rather than matrix multiplication by $W$. This creates an additive gradient path similar to skip connections in ResNets, so gradients flow easily across long sequences when $f = 1$.

Gated Recurrent Units (GRU) offer a lighter alternative by combining the cell state and hidden state into a single vector using reset and update gates. Neural Architecture Search (NAS) can also find domain-specific recurrent cell structures automatically using reinforcement learning.
