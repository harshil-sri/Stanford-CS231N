## Layers in CNNs

Conv layers slide 3D filters over the input to produce activation maps, keeping spatial layout intact. Pooling layers (max or average) downsample height and width for each feature map independently without using trainable parameters.

Batch Norm normalizes feature channels across the batch during training using minibatch stats, then applies learnable scale and shift parameters. At test time, it swaps minibatch stats for running averages of mean and variance. This turns BatchNorm into a simple linear operation that you can fold directly into adjacent layers.

Layer Norm normalizes across channels for a single sample. Instance Norm normalizes each channel individually per sample. Group Norm splits channels into groups and normalizes within each group per sample.


## CNN Architectures

Networks got much deeper over time, going from 8 layers in AlexNet to 152 layers in ResNet.

Very deep plain networks suffer from optimization problems where adding layers actually increases training error. ResNet solves this by having layers learn residual functions relative to their inputs.

Instead of fitting a mapping directly, a residual block fits $H(x) = F(x) + x$. If an identity mapping is optimal, the network can easily drive the residual part $F(x)$ to zero.


## Transfer Learning

When your dataset is small, pretrain a backbone on a large dataset like ImageNet, then transfer those learned features to your task.

If your dataset is small and looks similar to ImageNet, freeze the feature extractor and train only a new linear classifier on top. If you have a larger dataset, fine-tune the deeper layers or the whole network using a smaller learning rate, usually around a tenth of the pretraining rate.

If your dataset is small and structurally different from ImageNet, train a linear classifier on activations from middle layers instead of late layers. Early features are more generic and translate better across domains.


## Activation Functions

Sigmoid and Tanh squash inputs into bounded ranges ([0,1] and [-1,1]), but they saturate at both extremes, causing gradients to vanish during backpropagation.

ReLU avoids saturation for positive values and speeds up training, but negative inputs can kill neurons permanently when the gradient drops to zero.

Leaky ReLU and PReLU keep a small slope for negative inputs to prevent dead neurons. GELU smoothly scales inputs using the Gaussian CDF, helping modern models optimize better around zero.

## Data Preprocessing

For image data, preprocessing standardizes inputs by subtracting the per-channel mean and dividing by the per-channel standard deviation across the training set (yielding 3 mean and 3 std values for RGB images). You usually skip spatial whitening or PCA for vision models.


## Weight Initialization

If you initialize deep network weights too small, activation variances collapse to zero in deeper layers and backpropagated gradients vanish. If you make them too large, activations explode into saturated regions.

Xavier initialization sets the initial weight variance to $1 / \text{input size}$ to keep activation variance constant across layers for linear or tanh activations. It breaks down with ReLU because zeroing out negative outputs cuts the variance in half.

Kaiming (MSRA) initialization fixes this for ReLU by multiplying the variance factor by two: $W \sim \mathcal{N}(0, \sqrt{2 / D_{\text{in}}})$.

## Regularization & Training vs. Testing

A standard trick in neural network regularization is adding random noise during training and averaging it out at test time.

Inverted Dropout scales active activations by $1/p$ (where $p$ is the keep probability) during the forward pass in training. That keeps test time clean since you don't have to change anything at evaluation.


```python
# Inverted dropout implementation during training
mask = (np.random.rand(*H.shape) < p) / p
H *= mask
```


Data augmentation techniques (like random cropping, horizontal flipping, color jittering, and Cutout) add noise to the input during training to stop overfitting, without changing the ground-truth label.


## Hyperparameter Tuning Workflow

When tuning models, start by checking your initial loss, overfit a small batch to 100% accuracy, and then run a quick check across candidate learning rates (like 1e-1 down to 1e-4) to find values that drop the loss within ~100 iterations.

Watch your training and validation curves: a big gap between them means overfitting (add regularization or grab more data), while overlapping curves with bad accuracy mean underfitting (make the model bigger or train longer).

Use random search rather than grid search for hyperparameter optimization. Random sampling covers far more distinct values along key hyperparameter dimensions.
